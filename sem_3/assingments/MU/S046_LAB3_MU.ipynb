```python
data=[3,5,8,11,12,15]
sum=0
for i in data:
    sum=sum+i

mean=sum/len(data)
print(mean)


data.sort()
print(data)
n=len(data)
if n%2==0:
    median=(data[n//2-1]+data[n//2])/2
else:
    median=data[n//2]

print(median)

freq = {}

list(map(lambda x: freq.update({x: freq.get(x, 0) + 1}), data))

max_count = max(freq.values())

if max_count == 1:
    print("No mode")
else:
    mode = max(freq, key=freq.get)
    print("Mode:", mode)

        
        
```
