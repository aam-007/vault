```python
# q.1 Take one die, generate num 1 to 6, then generate occurances after n throws, and their probabilities. 
import random

n = int(input("Enter the number of throws: "))

occurrences = [0] * 6

for i in range(n):
    num = random.randint(1, 6)
    occurrences[num - 1] = occurrences[num - 1] + 1

print("Number\tOccurrences\tExperimental\tTheoretical")

for i in range(6):
    probability = occurrences[i] / n
    theoretical = 1 / 6

    print(i + 1, "\t", occurrences[i], "\t\t", probability, "\t\t", theoretical)
    
print("Num of throws: ", n)
```

```python
# q.2 take two die,number of trials, do same as above
import random

n = int(input("Enter the number of trials: "))

occurrences = [0] * 11  
for i in range(n):
    die1 = random.randint(1, 6)
    die2 = random.randint(1, 6)

    total = die1 + die2
    occurrences[total - 2] += 1

print("Sum\tOccurrences\tExperimental\tTheoretical")

ways = [1, 2, 3, 4, 5, 6, 5, 4, 3, 2, 1]

for i in range(11):
    total = i + 2

    probability = occurrences[i] / n
    theoretical = ways[i] / 36

    print(total, "\t", occurrences[i], "\t\t",
          probability, "\t\t", theoretical)


print("Num of throws: ", n)
```
