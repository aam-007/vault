```python
# Q1

rates = {
    "suv": 1000,
    "sedan": 5000,
    "hatchback": 250
}

car = input("Enter car (SUV/Sedan/Hatchback): ").lower()
days = int(input("Enter days: "))

deposit = 500

bill = rates[car] * days

if days > 10:
    bill = bill - (bill * 0.10)

final = bill - deposit

print("\n----- BILL -----")
print("Car:", car.upper())
print("Days:", days)
print("Bill:", bill)
print("Security Deposit:", deposit)
print("Final Amount:", final)
```

```python
# Q2
day = input("Enter day: ").lower()

if day in ["mon", "monday", "wed", "wednesday", "fri", "friday"]:
    car = "SUV"
    rate = 1000
elif day in ["tue", "tuesday", "thu", "thursday", "sat", "saturday"]:
    car = "Sedan"
    rate = 5000
elif day in ["sun", "sunday"]:
    car = "Hatchback"
    rate = 250
else:
    print("Invalid day")
    exit()

print("Car allotted:", car)
proceed = input("Proceed? (yes/no): ").lower()

if proceed != "yes":
    print("Booking cancelled.")
    exit()

days = 28
bill = rate * days

if days > 10:
    bill = bill - (bill * 0.10)

deposit = 500

choice = input("Adjust deposit? (yes/no): ").lower()

print("\n----- BILL -----")
print("Car:", car)
print("Bill:", bill)

if choice == "yes":
    print("Security Deposit:", deposit)
    print("Final Amount:", bill - deposit)
else:
    print("Bill Amount:", bill)
    print("Security Deposit:", deposit, "(Separate)")
    print("Total Payable:", bill + deposit)
```
